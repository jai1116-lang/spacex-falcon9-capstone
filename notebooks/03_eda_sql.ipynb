# 3. EDA with SQL

Loads the cleaned launch data into a SQLite table (IBM Db2 in the original
course environment) and answers a set of exploratory questions with SQL.

```python
import sqlite3
import pandas as pd

conn = sqlite3.connect(':memory:')
df = pd.read_csv('../data/spacex_launch_data.csv')
df.to_sql('SPACEXTBL', conn, if_exists='replace', index=False)
pd.read_sql("SELECT COUNT(*) as total_rows FROM SPACEXTBL", conn)
```

### Q1 — Unique launch sites

```python
pd.read_sql("SELECT DISTINCT LaunchSite FROM SPACEXTBL", conn)
```

### Q2 — Launches from CCAFS SLC-40

```python
pd.read_sql("""
SELECT COUNT(*) as ccafs_launches FROM SPACEXTBL
WHERE LaunchSite = 'CCAFS SLC 40'
""", conn)
```

### Q3 — Overall mission success rate

```python
pd.read_sql("""
SELECT
    SUM(Class) * 1.0 / COUNT(*) AS success_rate,
    SUM(Class) AS successes,
    COUNT(*) AS total
FROM SPACEXTBL
""", conn)
```

### Q4 — Launches to GTO (geosynchronous transfer orbit)

```python
pd.read_sql("""
SELECT COUNT(*) as gto_launches FROM SPACEXTBL WHERE Orbit = 'GTO'
""", conn)
```

### Q5 — Missions landed on a drone ship (ASDS)

```python
pd.read_sql("""
SELECT COUNT(*) as drone_ship_landings FROM SPACEXTBL WHERE Outcome = 'True ASDS'
""", conn)
```

### Q6 — Minimum payload mass

```python
pd.read_sql("SELECT MIN(PayloadMass) as min_payload_mass FROM SPACEXTBL", conn)
```

### Q7 — Total payload mass carried

```python
pd.read_sql("SELECT SUM(PayloadMass) as Total_Payload_Mass FROM SPACEXTBL", conn)
```

### Q8 — Mission outcome counts by launch site

```python
pd.read_sql("""
SELECT LaunchSite, COUNT(Outcome) as mission_outcome_count
FROM SPACEXTBL GROUP BY LaunchSite
""", conn)
```
